```python
import math

def hampiran_taylor(x, a=25, orde_maks=4):
    h = x - a
    koefisien = [math.sqrt(a)]

    for k in range(1, orde_maks + 1):
        sebelumnya = koefisien[-1]
        baru = sebelumnya * (0.5 - (k - 1)) / (k * a)
        koefisien.append(baru)

    hasil = []
    nilai_sebenarnya = math.sqrt(x)

    for n in range(orde_maks + 1):
        p = sum(koefisien[k] * h**k for k in range(n + 1))
        galat = abs(nilai_sebenarnya - p)
        hasil.append((n, p, galat))

    return hasil

x = 26
batas_galat = 1e-6
hasil = hampiran_taylor(x)
orde_minimal = None

print(f"Nilai sebenarnya sqrt({x}) = {math.sqrt(x):.9f}")

for orde, p, galat in hasil:
    print(f"P{orde}({x}) = {p:.9f}, galat = {galat:.9f}")
    if orde_minimal is None and galat < batas_galat:
        orde_minimal = orde

if orde_minimal is not None:
    print(f"Orde minimal agar galat < {batas_galat} adalah {orde_minimal}")
else:
    print("Orde maksimal belum memenuhi batas galat.")

```

```python

import math  # Menggunakan fungsi matematika bawaan Python.

def hampiran_taylor(x, a=25, orde_maks=4):  # Membuat fungsi untuk menghitung hampiran Taylor.
    h = x - a  # Menghitung selisih nilai x dengan titik pusat a.
    koefisien = [math.sqrt(a)]  # Menyimpan koefisien Taylor orde nol.

    for k in range(1, orde_maks + 1):  # Menghitung koefisien sampai orde maksimal.
        sebelumnya = koefisien[-1]  # Mengambil koefisien dari orde sebelumnya.
        baru = sebelumnya * (0.5 - (k - 1)) / (k * a)  # Menghitung koefisien Taylor berikutnya.
        koefisien.append(baru)  # Menambahkan koefisien ke dalam daftar.

    hasil = []  # Menyiapkan daftar untuk menyimpan hasil perhitungan.
    nilai_sebenarnya = math.sqrt(x)  # Menghitung nilai sebenarnya sebagai pembanding.

    for n in range(orde_maks + 1):  # Menghitung polinomial dari orde nol sampai orde maksimal.
        p = sum(koefisien[k] * h**k for k in range(n + 1))  # Menjumlahkan suku Taylor sampai orde n.
        galat = abs(nilai_sebenarnya - p)  # Menghitung galat absolut.
        hasil.append((n, p, galat))  # Menyimpan orde, nilai hampiran, dan galat.

    return hasil  # Mengembalikan seluruh hasil perhitungan.

x = 26  # Menentukan nilai x yang akan dihampiri.
batas_galat = 1e-6  # Menentukan batas galat sebesar 10 pangkat minus 6.
hasil = hampiran_taylor(x)  # Menghitung hampiran P0 sampai P4.
orde_minimal = None  # Menyiapkan variabel untuk menyimpan orde minimal.

print(f"Nilai sebenarnya sqrt({x}) = {math.sqrt(x):.9f}")  # Menampilkan nilai sebenarnya.

for orde, p, galat in hasil:  # Membaca hasil perhitungan setiap orde.
    print(f"P{orde}({x}) = {p:.9f}, galat = {galat:.9f}")  # Menampilkan nilai hampiran dan galat.
    if orde_minimal is None and galat < batas_galat:  # Memeriksa apakah galat sudah di bawah batas.
        orde_minimal = orde  # Menyimpan orde pertama yang memenuhi batas galat.

if orde_minimal is not None:  # Memeriksa apakah ditemukan orde yang memenuhi batas.
    print(f"Orde minimal agar galat < {batas_galat} adalah {orde_minimal}")  # Menampilkan orde minimal.
else:  # Berjalan jika tidak ada orde yang memenuhi batas galat.
    print("Orde maksimal belum memenuhi batas galat.")  # Menampilkan informasi batas galat belum tercapai.
```
